```python
import pandas as pd
data = {
    "student_ID": [101,102,103,104,105,106,107,108],
    "student": ["Asha","Ravi","priya","john","anu","kiran","ravi","meena"],
    "department": ["cse","cse","ece","ece","cse","eee","cse","eee"],
    "attendence": [90,75,95,60,80,70,75,85],
    "internalmarks": [85,70,90,55,78,65,70, None],
    "final_marks": [88,72,92,50,80,68,72,82]
    }
```

```python
df=pd.DataFrame(data)
df.to_csv("student.csv",index=False)
print("student.csv created succesfully")
```
